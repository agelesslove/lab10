```python
def task_7(input_val=None):
    """
    Алгоритм переведення чисел (цілих та з рухомою комою) 
    з десяткової системи числення в двійкову (Варіант 7).
    """
    if input_val is None:
        input_str = input("Введіть число: ")
    else:
        input_str = str(input_val)

    # Контроль коректності вводу
    try:
        num = float(input_str)
    except ValueError:
        return "Помилка. У списку присутнє нечислове значення або введено нечисловий символ"

    is_negative = num < 0
    num = abs(num)

    int_part = int(num)
    frac_part = num - int_part

    # Переведення цілої частини
    if int_part == 0:
        bin_int = "0"
    else:
        bin_int = ""
        temp_int = int_part
        while temp_int > 0:
            bin_int = str(temp_int % 2) + bin_int
            temp_int //= 2

    # Переведення дробової частини (до 10 знаків)
    bin_frac = ""
    if frac_part > 0:
        temp_frac = frac_part
        while temp_frac > 0 and len(bin_frac) < 10:
            temp_frac *= 2
            bit = int(temp_frac)
            bin_frac += str(bit)
            temp_frac -= bit

    result = bin_int + ('.' + bin_frac if bin_frac else '')
    if is_negative:
        result = '-' + result

    return f"'{input_str} у двійковій системі буде {result}'"
```
